```python
from firewxpy import rtma_conus
from wxdata import rtma
```

### Part 1

We will now create our RTMA Temperature graphic for CONUS. In this example, we will download and plot the data all within the FireWxPy function.

Since the default temperature range is -30F to 130F, let's change the range to get a better color contrast. 
We will use 40F to 110F.

```python
rtma_conus.plot_temperature(start=40,
                           stop=110)
```

#### Using WxData to Download our RTMA Data

```python
ds = rtma()
```

```python
ds
```

#### Creating Our Variety of RTMA Graphics

Since we used WxData to convert our temperature from Kelvin to Fahrenheit, we will set `convert_temperature=False` for all functions involving temperature. 

```python
rtma_conus.plot_temperature(start=40,
                            stop=110,
                            ds=ds,
                            filename="RTMA Temperature CONUS States & Counties.png",
                            convert_temperature=False)

rtma_conus.plot_temperature(start=40,
                            stop=110,
                            ds=ds,
                            filename="RTMA Temperature CONUS GACC & PSA.png",
                            reference_system='GACC & PSA',
                            convert_temperature=False,
                            show_states=False,
                            show_counties=False,
                            show_gacc_boundaries=True,
                            show_predictive_services_areas=True,
                            colormap='custom')

rtma_conus.plot_temperature(start=40,
                            stop=110,
                            ds=ds,
                            filename="RTMA Temperature CONUS NWS CWA & Public Zones.png",
                            reference_system='NWS CWAs & Public Zones',
                            convert_temperature=False,
                            show_states=False,
                            show_counties=False,
                            show_nws_cwa=True,
                            show_nws_public_zones=True,
                            colormap='custom',
                            colors=['rebeccapurple',
                                    'orchid',
                                    'navy',
                                    'blue',
                                    'lightseagreen',
                                    'gold',
                                    'goldenrod',
                                    'darkorange',
                                    'crimson',
                                    'darkred'])

rtma_conus.plot_temperature(start=40,
                            stop=110,
                            ds=ds,
                            filename="RTMA Temperature CONUS NWS CWA & Fire Weather Zones.png",
                            reference_system='NWS CWAs & Fire Weather Zones',
                            convert_temperature=False,
                            show_states=False,
                            show_counties=False,
                            show_nws_cwa=True,
                            show_nws_fire_weather_zones=True,
                            colormap='custom',
                            colors=['rebeccapurple',
                                    'orchid',
                                    'navy',
                                    'blue',
                                    'lightseagreen',
                                    'gold',
                                    'goldenrod',
                                    'darkorange',
                                    'crimson',
                                    'darkred'])
```

#### Creating Our Regional Graphics For A Variety of Regions

In this example, we will create plots for various parameters for various regions with various reference systems.

```python
# Since this image has a narrower domain in the horizontal (W-E), let's set signature_text_new_lines=True.
rtma_conus.plot_temperature(region='ca',
                            start=50,
                            stop=110,
                            ds=ds,
                            filename='RTMA Temperature California States & Counties',
                            convert_temperature=False,
                            colorbar_shrink=0.8,
                            decimate=20,
                            signature_text_new_lines=True,
                            signature_textbox_y_position=-0.12,
                            pixel_query_value_fontsize=8)

# Since this image has a narrower domain in the horizontal (W-E), let's set signature_text_new_lines=True.
rtma_conus.plot_wind_speed(region='nm',
                            start=10,
                            stop=50,
                            ds=ds,
                            filename='RTMA Wind Speed New Mexico NWS CWA & Public Zones.png',
                            reference_system='NWS CWAs & Public Zones',
                            show_states=False,
                            show_counties=False,
                            show_nws_cwa=True,
                            show_nws_public_zones=True,
                            decimate=10,
                            colorbar_shrink=0.75,
                            signature_text_new_lines=True,
                            signature_textbox_y_position=-0.12)

# Since the converted wind speed is kept in the dataset after converting from m/s to mph the first time
# Let's set convert_wind_speed=False. 
rtma_conus.plot_relative_humidity_and_wind(region='co',
                                           ds=ds,
                                           filename='RTMA RH & Wind Colorado States & Counties',
                                           decimate=20,
                                           convert_wind_speed=False)

# Here, let's change the line thicknesses and colors of the boundaries
rtma_conus.plot_dew_point_depression_and_gust(region='oscc',
                                              ds=ds,
                                              filename='RTMA Dew Point Depression & Wind OSCC GACC & CalFire.png',
                                              reference_system='GACC & CalFire',
                                              show_states=False,
                                              show_gacc_boundaries=True,
                                              show_calfire_boundaries=True,
                                              gacc_border_linewidth=1,
                                              calfire_boundary_linewidth=0.5,
                                              calfire_boundary_color='crimson',
                                              decimate=20,
                                              signature_textbox_y_position=-0.1)
```

##### Defining A Region Using Custom Boundaries

In this example, we will plot the latest RTMA for relative humidity and wind across the south central U.S. to analyze the dryline. 

```python
rtma_conus.plot_relative_humidity_and_wind(region='custom',
                                          ds=ds,
                                          filename='RTMA Dry Line.png',
                                          western_bound=-107.3,
                                          eastern_bound=-96.5,
                                          southern_bound=29.25,
                                          northern_bound=37.4,
                                          decimate=20,
                                          convert_wind_speed=False,
                                          signature_textbox_y_position=-0.1)
```
