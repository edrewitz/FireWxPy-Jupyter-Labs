In this example, we will plot the following RTMA Comparisons:

1) RTMA 24-Hour Temperature Comparison for CONUS.
2) RTMA 1-Hour Dew Point Depression & Gust Comparison for the South Ops GACC region.
3) RTMA 1-Hour Relative Humidity & Gust Comparison for the South Ops GACC region.

## Imports

We will import FireWxPy and WxData. We will use WxData to download our 1 hourly change data and pass in our two datasets `ds1` and `ds2` into the plotting functions.

```python
from firewxpy import rtma_comparison_conus
from wxdata import rtma_comparison
```

### Creating Our RTMA 24-Hour Temperature Comparison for CONUS.

```python
rtma_comparison_conus.plot_temperature()
```

### Using WxData to Download our RTMA Datasets and FireWxPy to Plot Our 1-Hourly Change Graphics.

```python
ds_current, ds_comparison = rtma_comparison(hours=1)
```

```python
rtma_comparison_conus.plot_dew_point_depression_and_gust(region='oscc',
                                                         start=-5,
                                                         stop=5,
                                                         decimate=20,
                                                         hours=1,
                                                         show_states=False,
                                                         show_gacc_boundaries=True,
                                                         show_predictive_services_areas=True,
                                                         primary_title_fontsize=8,
                                                         secondary_title_fontsize=7,
                                                         ds1=ds_current,
                                                         ds2=ds_comparison,
                                                         filename='OSCC 1 Hourly DD & Gust Change.png',
                                                         reference_system='GACC & PSA & Counties',
                                                         colorbar_interval=1,
                                                         signature_textbox_y_position=-0.1,
                                                         gacc_border_linewidth=1,
                                                         predictive_services_areas_linewidth=0.5)
```

<img src="https://github.com/edrewitz/firewxpy/blob/main/Example%20Graphics/RTMA%20Comparison/OSCC%201%20Hourly%20DD%20&%20Gust%20Change.png?raw=true" width="1000" height="1000" />

```python
# Since the converted wind speed is kept in the dataset after converting from m/s to mph the first time
# Let's set convert_wind_speed=False. 
rtma_comparison_conus.plot_relative_humidity_and_gust(region='oscc',
                                                      start=-5,
                                                      stop=5,
                                                      decimate=20,
                                                      hours=1,
                                                      show_states=False,
                                                      show_gacc_boundaries=True,
                                                      show_predictive_services_areas=True,
                                                      primary_title_fontsize=8,
                                                      secondary_title_fontsize=7,
                                                      ds1=ds_current,
                                                      ds2=ds_comparison,
                                                      filename='OSCC 1 Hourly RH & Gust Change.png',
                                                      reference_system='GACC & PSA & Counties',
                                                      convert_wind_speed=False,
                                                      colorbar_interval=1,
                                                      signature_textbox_y_position=-0.1,
                                                      gacc_border_linewidth=1,
                                                      predictive_services_areas_linewidth=0.5)
```
