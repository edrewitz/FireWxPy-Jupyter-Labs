# Creating Observed Soundings In FireWxPy

In this example, we will use FireWxPy to create the following:

1) A graphic of the current sounding from San Diego, CA (NKX)
2) A graphic of the sounding from the El Reno Tornado on May 31st, 2013 (OUN)
3) A graphic of the OKX sounding during the Valentine's Day Ice Storm of 2007 (My favorite storm since this was my first ice storm and I was a 7th grader who grew up on Long Island).

[Documentation](https://github.com/edrewitz/firewxpy/blob/main/Documentation/firewxpy%202.0/observed%20soundings.md#observed-soundings)

## Imports

```python
from firewxpy import observed_soundings
```

### Current NKX Sounding

In this first example, we will plot the latest NKX Sounding with all the default settings.

There are many settings for increased customization beginning here in FireWxPy 2.0. 

Please see the documentation and following examples for more information on customization. 

```python
observed_soundings.plot_observed_sounding('nkx')
```

![image](https://github.com/edrewitz/firewxpy/blob/main/Example%20Graphics/NKX.png?raw=true)

### El Reno Tornado Sounding (OUN May 31st, 2013 18z)

In this example, we want to plot an archived sounding. 

I will also make a custom dark mode in this example and change around a medley of colors. 

```python
observed_soundings.plot_observed_sounding('oun',
                                          current=False,
                                          custom_time='2013-05-31:18',
                                          temperature_line_color='pink',
                                          dew_point_line_color='greenyellow',
                                          wind_barb_color_table='cool',
                                          lcl_marker_color='gold',
                                          skew_t_background_color='black',
                                          facecolor='gainsboro')
```

![image](https://github.com/edrewitz/firewxpy/blob/main/Example%20Graphics/OUN.png?raw=true)

### Long Island/NYC Valentine's Day Ice Storm of 2007 (OKX February 14th, 2007 12z

```python
observed_soundings.plot_observed_sounding('okx',
                                          current=False,
                                          custom_time='2007-02-14:12',
                                          wind_barb_color_table='copper',
                                          lcl_marker_color='forestgreen',
                                          skew_t_background_color='white')
```

![image](https://github.com/edrewitz/firewxpy/blob/main/Example%20Graphics/OKX.png?raw=true)
