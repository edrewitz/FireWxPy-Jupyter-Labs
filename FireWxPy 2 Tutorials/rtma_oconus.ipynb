## Imports

***Important Thing To Note***

*We will use WxData to download the data outside of the function for the 3 Alaska graphics and pass them in since they are using the same dataset.*

Users can also download and plot the data inside of the functions. *Recommended for a single graphic or using different regions (CONUS vs. Alaska vs. Hawaii)

```python
from firewxpy import(
    rtma_alaska,
    rtma_hawaii
)
from wxdata import rtma
```

### Using WxData to retrieve our AK RTMA Data.

Since the default setting is `model='rtma'`, we need to change it to `model='ak rtma'`.

```python
ds = rtma(model='ak rtma')
```

### Creating our various AK RTMA Graphics

```python
rtma_alaska.plot_temperature(start=30,
                             stop=70,
                             ds=ds,
                             filename='AK RTMA Temperature Alaska.png',
                             convert_temperature=False)
                             
```

<img src="https://github.com/edrewitz/firewxpy/blob/main/Example%20Graphics/RTMA/AK%20RTMA%20Temperature%20Alaska.png?raw=true" width="1000" height="1000" />

```python
rtma_alaska.plot_temperature_and_wind(region='aer',
                                      ds=ds,
                                      start=30,
                                      stop=70,
                                      filename='AK RTMA AER Domain.png',
                                      decimate=20)
```

<img src="https://github.com/edrewitz/firewxpy/blob/main/Example%20Graphics/RTMA/AK%20RTMA%20AER%20Domain.png?raw=true" width="1000" height="1000" />

```python
# Since the converted wind speed is kept in the dataset after converting from m/s to mph the first time
# Let's set convert_wind_speed=False. 
rtma_alaska.plot_temperature_and_wind(region='alu',
                                      ds=ds,
                                      start=30,
                                      stop=70,
                                      filename='AK RTMA ALU Domain.png',
                                      decimate=50,
                                      convert_wind_speed=False)
```

### Creating our RTMA Wind Gust Graphic for Hawaii

```python
rtma_hawaii.plot_wind_gust(filename='RTMA Wind Gust Hawaii.png',
                           start=10)
```
