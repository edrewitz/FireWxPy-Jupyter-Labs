## Imports

We will import FireWxPy and WxData. We will use WxData to download our 1 hourly change data and pass in our two datasets `ds1` and `ds2` into the plotting functions.

```python
from firewxpy import(
    rtma_comparison_alaska,
    rtma_comparison_hawaii
)
from wxdata import rtma_comparison
```

### Creating Our RTMA 24-Hour Temperature Comparison for CONUS.

```python
rtma_comparison_alaska.plot_temperature(filename='AK RTMA Temperature Comparison')
```

### Using WxData to Download our RTMA Datasets and FireWxPy to Plot Our 1-Hourly Change Graphics.

```python
ds_current, ds_comparison = rtma_comparison(model='ak rtma',
                                            hours=1)
```

```python
rtma_comparison_alaska.plot_temperature_and_wind(region='aer',
                                      ds1=ds_current,
                                      ds2=ds_comparison,
                                      start=-5,
                                      stop=5,
                                      filename='AK RTMA Comparison AER Domain.png',
                                      decimate=20,
                                      colorbar_interval=1)
```

<img src="https://github.com/edrewitz/firewxpy/blob/main/Example%20Graphics/RTMA%20Comparison/AK%20RTMA%20Comparison%20AER%20Domain.png?raw=true" width="1000" height="1000" />

```python
# Since the converted wind speed is kept in the dataset after converting from m/s to mph the first time
# Let's set convert_wind_speed=False. 
rtma_comparison_alaska.plot_temperature_and_wind(region='alu',
                                      ds1=ds_current,
                                      ds2=ds_comparison,
                                      start=-5,
                                      stop=5,
                                      filename='AK RTMA Comparison ALU Domain.png',
                                      decimate=50,
                                      convert_wind_speed=False,
                                      colorbar_interval=1)
```
