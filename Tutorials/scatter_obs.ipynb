```python
'''
In this example we will make plots showing the relative humidity, temperature and wind speed observations in a scatterplot form
We will make a medley of these plots. We will make 1 set for CONUS and another set for New York State (the state I was born and raised in). 

Since we are making a medley of plots, let's download the data outside the function and pass it in.

First let's start with our imports. 

'''
from firewxpy import obs
from firewxpy import scatter_obs
```

```python
'''
Now we need to get our data. To do this we will use the obs class of FireWxPy
'''
df, time = obs.get_metar_data()
```

```python
'''
We now need to use our functions in the scatter_obs class to make our graphics.

We need to set data=True and pass in the Pandas DataFrame and the time of the DataFrame as well.

For the second set of plots, we need to set state='ny'

Now let's make our graphics!
'''
scatter_obs.plot_relative_humidity(data=True, df=df, time=time)
scatter_obs.plot_temperature(data=True, df=df, time=time)
scatter_obs.plot_wind_speed(data=True, df=df, time=time)

scatter_obs.plot_relative_humidity(data=True, df=df, time=time, state='ny')
scatter_obs.plot_temperature(data=True, df=df, time=time, state='ny')
scatter_obs.plot_wind_speed(data=True, df=df, time=time, state='ny')
```
