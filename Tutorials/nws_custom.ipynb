```python
'''
In this example, we will make a set of NWS Max RH Forecast Trend and NWS Min RH Forecast Trend graphics.

We will also make this for an area of custom bounds.

Since we are also making a couple of different plots, let's show you how to download the data from outside and pass it in. 

Let's start with our imports:

'''
from firewxpy import NDFD_GRIDS
from firewxpy import nws_relative_humidity_forecast as rh
```

```python
'''
We need to grab our data. To do this we need to use our NDFD_GRIDS class
The get_ndfd_grids_xarray() takes 3 required arguments:

1) directory_name (String) - The directory name. See documentation for various directory paths. 

Docs: https://github.com/edrewitz/firewxpy/blob/main/Documentation/miscellaneous.md#get_ndfd_grids_xarray

2) parameter (String) - The parameter we want
3) state (String) - In this example, I will use CONUS. 

Our function returns the dataset for the short-term grids (ds_short) and the extended grids (ds_extended)
'''
maxrh_short, maxrh_extended = NDFD_GRIDS.get_ndfd_grids_xarray('/SL.us008001/ST.opnl/DF.gr2/DC.ndfd/AR.conus/', 'ds.maxrh.bin', 'conus')
minrh_short, minrh_extended = NDFD_GRIDS.get_ndfd_grids_xarray('/SL.us008001/ST.opnl/DF.gr2/DC.ndfd/AR.conus/', 'ds.minrh.bin', 'conus')
```

```python
'''
Since we are making a plot with custom boundaries, we need to do the following:

1) set state='custom'
2) define our plot boundaries
3) pass in our datasets
4) play around with the x,y coordinates for the textboxes and shrink factor of the colorbar. 
   In this example, the default x,y values worked well but we needed to increase our shrink
   factor from the default 0.7 to 0.9 to get our colorbar looking nice. 

'''

rh.plot_maximum_relative_humidity_forecast_trend(western_bound=-110, eastern_bound=-90, southern_bound=27, northern_bound=47, data=True, ds_short=maxrh_short, ds_extended=maxrh_extended, state='custom', shrink=0.9)
rh.plot_minimum_relative_humidity_forecast_trend(western_bound=-110, eastern_bound=-90, southern_bound=27, northern_bound=47, data=True, ds_short=minrh_short, ds_extended=minrh_extended, state='custom', shrink=0.9)
```
