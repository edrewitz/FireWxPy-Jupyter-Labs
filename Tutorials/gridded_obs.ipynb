```python
'''
In this example we will grid the Relative Humidity, Temperature and Wind Observations and make a plot of the gridded data. 
We also will have the values of some of the observations overlayed onto the graphic. 

First, let's start with our imports:
'''

from firewxpy import obs
from firewxpy import gridded_obs
```

```python
'''
In this example we will follow our best practice of downloading the data outside of the function 
to limit data requests on the server where the data is stored.

I also want to look at a Geographic Area Coordination Center's geographic area in this example. 
Let's use my office's (South Ops) area of responsiblity. 

I will first want to pass in a value for gacc_region.
Since I want South Ops, I will set gacc_region='oscc'
Please see documentation for other gacc_region abbreviations. 

I also want to change my reference system to look at GACC & PSA & Counties. 
As a Predictive Services Meteorologist, I look at the boundaries of the Geographic Area and the Predictive Services Areas. 
I also like to have the counties reference lurking in the back as well as an additional reference point. 

Thus, let's change our reference_system from the default value of 'States & Counties' to 'GACC & PSA & Counties'.

Next, we need to pass in our data, so we need to:
1) set data=True
2) set df=df
3) set time=time

Last, we need to set our mask. This has to do with how far spaced the observations are. 
Default value is 300000 which works well with CONUS. However, when looking at a smaller area,
I want more observations sites to be displayed. How we do that is we need to set our mask
value to a lower value. 60000 works well for the South Ops domain. To do this we need to 
set mask=60000. 

Now let's make our graphics!

'''

df, time = obs.get_metar_data()
```

```python
gridded_obs.plot_relative_humidity_observations(gacc_region='oscc', reference_system='GACC & PSA & Counties', data=True, df=df, time=time, mask=60000)
gridded_obs.plot_temperature_observations(gacc_region='oscc', reference_system='GACC & PSA & Counties', data=True, df=df, time=time, mask=60000)
gridded_obs.plot_wind_observations(gacc_region='oscc', reference_system='GACC & PSA & Counties', data=True, df=df, time=time, mask=60000)
```
