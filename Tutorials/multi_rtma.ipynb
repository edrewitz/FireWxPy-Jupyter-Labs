```python
'''
In this example, we will make a medley of RTMA Graphics across Alaska Region. We will make 24-Hour Wind Speed Comparisons and 24-Hour RH Comparisons.

Due to us making a medley of graphics using the same dataset, we will follow best practice and download the data outside of the function using the 
RTMA class in the firewxpy.data_access module. 

This is the best practice because this limits the amount of data requests on the servers where the data is being stored. 

Now let's import our rtma module and RTMA class from our firewxpy.data_access module as well as our plot_creation_time() function

'''
from firewxpy import rtma
from firewxpy import RTMA
from firewxpy import plot_creation_time
```

```python
'''
In this cell we will download our dataset from the NOAA/NCEP/NOMADS Server. 

We first need our current time in UTC. 

We will also set our region to 'AK'
'''

local_time, utc_time = plot_creation_time()

ds, ds_24, time, time_24 = RTMA.get_rtma_datasets('AK', utc_time)
```

```python
'''
Now it's time for us to make our plot. You will see FireWxPy build a new branch to the Weather Data directory since this will be our first time
running these plotting functions since our first time user set up. 

Since we downloaded the data outside of the function, we need to change more optional arguments:

1) We need to set data=True
2) We need to set ds=ds, ds_24=ds_24, time=time, time_24=time_24
3) For any specific CWA we need to set the cwa to the abbreviation

NWS Anchorage East Domain: cwa='aer'
NWS Anchorage West Domain: cwa='alu'
NWS Juneau: cwa='ajk'
NWS Fairbanks: cwa='afg'
'''
# 24-Hour Relative Humidity Comparisons
rtma.plot_24_hour_relative_humidity_comparison(state='ak', data=True, ds=ds, ds_24=ds_24, time=time, time_24=time_24)
rtma.plot_24_hour_relative_humidity_comparison(state='ak', data=True, ds=ds, ds_24=ds_24, time=time, time_24=time_24, cwa='aer')
rtma.plot_24_hour_relative_humidity_comparison(state='ak', data=True, ds=ds, ds_24=ds_24, time=time, time_24=time_24, cwa='alu')
rtma.plot_24_hour_relative_humidity_comparison(state='ak', data=True, ds=ds, ds_24=ds_24, time=time, time_24=time_24, cwa='ajk')
rtma.plot_24_hour_relative_humidity_comparison(state='ak', data=True, ds=ds, ds_24=ds_24, time=time, time_24=time_24, cwa='afg')

# 24-Hour Temperature Comparisons
rtma.plot_24_hour_temperature_comparison(state='ak', data=True, ds=ds, ds_24=ds_24, time=time, time_24=time_24)
rtma.plot_24_hour_temperature_comparison(state='ak', data=True, ds=ds, ds_24=ds_24, time=time, time_24=time_24, cwa='aer')
rtma.plot_24_hour_temperature_comparison(state='ak', data=True, ds=ds, ds_24=ds_24, time=time, time_24=time_24, cwa='alu')
rtma.plot_24_hour_temperature_comparison(state='ak', data=True, ds=ds, ds_24=ds_24, time=time, time_24=time_24, cwa='ajk')
rtma.plot_24_hour_temperature_comparison(state='ak', data=True, ds=ds, ds_24=ds_24, time=time, time_24=time_24, cwa='afg')
```
