```python
'''
In this example we will create an RTMA graphic with custom boundaries. 

We will look at the RTMA Wind Speed over the Northern Rockies.
To make it more fun, we will also look at the 24-Hour RTMA Wind Speed Comparison and
download the data outside of the function and pass it in since we don't need to send
2 seperate data requests to the server. 

First let's start with our imports
'''
from firewxpy import rtma
from firewxpy import RTMA
from firewxpy import plot_creation_time
```

```python
'''
In this cell we will download our dataset from the NOAA/NCEP/NOMADS Server. 

We first need our current time in UTC. 

We will also set our region to 'conus'
'''

local_time, utc_time = plot_creation_time()

ds, ds_24, time, time_24 = RTMA.get_rtma_datasets('conus', utc_time)
```

```python
'''
Since we are making a plot using custom boundaries, we need to change around more optional arguments:

1) We need to define the boundaries of the plot in decimal degree latitude/longitude.
2) We also might need to play around with x1, y1, x2, y2, x3, y3 and shrink
    *In this case all the default x and y coordinates look good for the locations of the text boxes*
    *We had to decrease the shrink to make the colorbar smaller so it can line up better with our plot*

*We also had to pass in the optional arguments for when downloading the data outside of the function and passing it in*

Now let's make our graphics. 
'''

rtma.plot_wind_speed(state='custom', western_bound=-117, eastern_bound=-100, southern_bound=40, northern_bound=50, data=True, ds=ds, time=time, shrink=0.5)
rtma.plot_24_hour_wind_comparison(state='custom', western_bound=-117, eastern_bound=-100, southern_bound=40, northern_bound=50, data=True, ds=ds, ds_24=ds_24, time=time, time_24=time_24, shrink=0.5)
```
