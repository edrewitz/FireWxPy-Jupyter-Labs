```python
'''
In this example we will create RTMA Critical Fire Weather plots using the latest hourly data (19z) for the following areas:

1) Custom boundaries looking at the Mid-Atlantic & SE
2) MD
3) VA
4) NC
5) SC
6) GA

We will also change our default wind and relative humidity thresholds as the defaults are set to Southern California standards
and the Southeast uses different threshold standards. 

We will change these standards to fit the criteria used in the text of the NWS RFW
Wind Speed >= 15 MPH & RH <= 30%

(Default values are RH <= 15% & Wind Speed >= 25 MPH)

First let's start with our imports
'''
from firewxpy import rtma
from firewxpy import RTMA
from firewxpy import plot_creation_time
```

```python
'''
In this cell we will download our dataset from the NOAA/NCEP/NOMADS Server. 

We first need our current time in UTC. 

We will also set our region to 'conus'
'''

local_time, utc_time = plot_creation_time()

ds, ds_24, time, time_24 = RTMA.get_rtma_datasets('conus', utc_time)
```

```python
'''
Now let's make our plots!
'''
rtma.plot_critical_firewx(state='custom', western_bound=-85.47, eastern_bound=-75.5, southern_bound=32.85, northern_bound=40, data=True, ds=ds, time=time, low_rh_threshold=30, high_wind_threshold=15)
rtma.plot_critical_firewx(state='md', data=True, ds=ds, time=time, low_rh_threshold=30, high_wind_threshold=15)
rtma.plot_critical_firewx(state='va', data=True, ds=ds, time=time, low_rh_threshold=30, high_wind_threshold=15)
rtma.plot_critical_firewx(state='nc', data=True, ds=ds, time=time, low_rh_threshold=30, high_wind_threshold=15)
rtma.plot_critical_firewx(state='sc', data=True, ds=ds, time=time, low_rh_threshold=30, high_wind_threshold=15)
rtma.plot_critical_firewx(state='ga', data=True, ds=ds, time=time, low_rh_threshold=30, high_wind_threshold=15)
```
